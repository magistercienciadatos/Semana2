# **I. `Scope` y los `Namespaces`**

Antes de abordar el concepto de clases, resulta fundamental comprender los principios de scope y namespace en Python, ya que ambos determinan cómo se organizan, resuelven y acceden los nombres dentro de un programa.

Un namespace puede entenderse como una estructura de mapeo entre identificadores y objetos. En otras palabras, constituye el contexto en el que un nombre queda asociado a una referencia específica dentro del entorno de ejecución. Los namespaces se generan en distintos momentos del ciclo de vida del programa y pueden coexistir con diferentes alcances y duraciones, según la naturaleza del bloque de código en el que se originan.

Durante la ejecución de un programa en Python 3 se crean múltiples namespaces, cada uno vinculado a determinados niveles de scope, lo que define desde qué regiones del código dichos nombres pueden ser accedidos o utilizados. Comprender esta relación es esencial para interpretar correctamente la visibilidad de las variables, evitar ambigüedades en la resolución de nombres y diseñar estructuras de programación más robustas y mantenibles.

# **1. ¿Que es un `Scope`?**

El scope (ámbito) define desde dónde una variable puede ser utilizada.

````python
def saludar():
    mensaje = "Hola"
    print(mensaje)

saludar()
print(mensaje)
````

La variable mensaje existe solo dentro de la función porque su scope es local.

# **2. ¿Que es un Namespace?**

Un namespace es un espacio donde Python almacena nombres y sus referencias a objetos.
El namespace responde a la pregunta: ¿dónde está almacenado este nombre?

# **3. Relación con Programación Orientada a Objetos**

En Python, las clases y los objetos tienen su propio namespace.


````python
class Persona:
    especie = "Humano"
````

La clase almacena:

Persona

 └── especie

````python
p = Persona()
p.nombre = "Ana"
````

¿Cómo busca Python los atributos?

````python
#Cuando escribimos:
p.especie
````

Python busca:

1. En el namespace del objeto p.

2. Si no lo encuentra, en el namespace de la clase Persona.

3. Luego en las clases padre (si existen).

````python
class Persona:

    def __init__(self, nombre):
        self.nombre = nombre
````

self.nombre guarda el atributo dentro del namespace del objeto.

Por eso

````python
p = Persona("Ana")
print(p.nombre)
````

El namespace es donde se guardan los nombres; el scope determina desde dónde pueden utilizarse. En POO, cada clase y cada objeto tienen su propio namespace, y Python busca atributos y métodos recorriendo esos namespaces.

# **II. Programación Orientación a Objetos**

La orientación a objetos es un paradigma de programación que permite modelar y organizar soluciones computacionales mediante entidades denominadas objetos, las cuales integran datos y comportamientos relacionados dentro de una misma unidad lógica. Este enfoque favorece una estructuración más modular, reutilizable y comprensible del software, al permitir agrupar propiedades y funcionalidades afines en abstracciones coherentes.

Python es un lenguaje orientado a objetos, y en él todas las estructuras de datos y componentes manipulados durante la ejecución son tratados como objetos. Esto implica que tanto los tipos básicos como las estructuras más complejas poseen identidad, estado y comportamiento asociados.

En este paradigma, la construcción fundamental para definir nuevas abstracciones es la clase, la cual actúa como un modelo o plantilla que describe los atributos y métodos que tendrán los objetos creados a partir de ella. A su vez, los objetos constituyen instancias concretas de una clase y permiten representar entidades, procesos o conceptos del mundo real dentro de un sistema computacional. De este modo, la programación orientada a objetos facilita el diseño de programas más expresivos, escalables y alineados con problemas de mayor complejidad, como los abordados en ciencia de datos e inteligencia artificial.

### 1. Clases

Las clases son el elemento central de la programación orientada a objetos, ya que permiten organizar en una misma estructura tanto los datos como las funciones asociadas a ellos. Cuando se define una clase en Python, se genera un nuevo tipo de dato, a partir del cual pueden crearse distintas instancias, conocidas como objetos. Cada objeto puede almacenar información propia mediante atributos, que representan su estado, y también puede disponer de métodos, es decir, funciones definidas dentro de la clase que permiten manipular o modificar dicho estado.

La sintaxis empleada para definir clases en Python 3 es la siguiente:



### 2. Metodos

Dentro de una clase pueden definirse funciones, y cuando una función pertenece a una clase se denomina método. Los métodos representan los comportamientos o acciones de los objetos. En general, todo lo visto sobre funciones también se aplica a los métodos; la principal diferencia es que estos se invocan a través de un objeto o de la clase a la que pertenecen.

````python
class Moto:
    def velocidad_maxima():
        """Este método devuelve la velocidad máxima de la moto."""
        print("Velocidad máxima: ???")
````

### 3. Parámetro `self`

Al definir métodos dentro de una clase, es necesario incluir el parámetro self como primer argumento. Este parámetro representa la instancia concreta del objeto que está utilizando el método en ese momento. Es decir, self permite que el método sepa sobre qué objeto está actuando.

Gracias a self, un método puede acceder a los atributos propios de cada objeto y también invocar otros métodos definidos en la misma clase. Esto es fundamental porque, aunque varios objetos provengan de una misma clase, cada uno puede almacenar valores distintos en sus atributos. Por ello, self permite distinguir la información particular de cada instancia.

Por ejemplo, si una clase Persona tiene un atributo nombre, al escribir self.nombre se está haciendo referencia al nombre del objeto específico que ejecuta el método, no al de la clase en general ni al de otro objeto. De esta manera, cada instancia puede trabajar con sus propios datos de forma independiente.

En síntesis, self es necesario porque conecta los métodos con los datos del objeto que los ejecuta, permitiendo que cada instancia mantenga y manipule su propia información dentro de la clase.

````python
class Moto:
    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima: ???")
````

### 5. Atributos

Dentro de una clase también pueden definirse variables, las cuales, en el contexto de la programación orientada a objetos, reciben el nombre de atributos. Estos atributos permiten almacenar información relacionada con los objetos de la clase.

En Python, los atributos se clasifican en dos tipos principales: atributos de clase y atributos de instancia. Los primeros pertenecen a la clase en general y son compartidos por todos los objetos creados a partir de ella, mientras que los segundos corresponden a cada objeto en particular, por lo que cada instancia puede tener sus propios valores.

Un atributo de clase es una variable definida dentro de la clase que pertenece a la propia clase y, por lo tanto, es compartida por todos los objetos creados a partir de ella. Esto significa que su valor no es exclusivo de una instancia en particular, sino común para todas. Para acceder al valor de un atributo de clase, se utiliza la siguiente sintaxis:

```
<nombre_objeto>.<nombre_atributo_clase>
```

````python
class Moto:
    atributo_clase = 150

    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima:", self.atributo_clase)
````

````python
moto1 = Moto()
````

````python
moto1.velocidad_maxima()
````

````python
moto1.atributo_clase
````

````python
moto2 = Moto()
````

````python
moto2.velocidad_maxima()
````

Un atributo de instancia es una variable que pertenece a un objeto específico, por lo que su valor puede ser distinto en cada instancia de la clase. Estos atributos se definen generalmente dentro del método especial __init__(), conocido como constructor, que se ejecuta al crear un objeto. Para acceder al valor de un atributo de instancia, se utiliza la siguiente sintaxis:

```
<nombre_objeto>.<nombre_atributo_instancia>
```

### 6. Método `__init__()`

El método `__init__()` es un método especial que Python ejecuta automáticamente cada vez que se crea una nueva instancia de una clase. Su nombre tiene dos guiones bajos al inicio y dos al final, ya que forma parte de los métodos especiales del lenguaje, los cuales siguen una convención específica para distinguirlos de los métodos definidos por el usuario.

Este método se conoce como constructor porque permite inicializar el objeto recién creado, asignándole valores a sus atributos. De esta manera, cada instancia puede quedar configurada con información propia desde el momento en que es creada.

````python
class Moto:

    def __init__(self, vel_max, consumo_medio):
        self.vel_max = vel_max
        self.con_medio = consumo_medio

    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima:", self.vel_max)

    def consumo_medio(self):
        print("El consumo medio es:", self.con_medio)
````

### 7. Herencia

Es un mecanismo de la programación orientada a objetos que permite crear nuevas clases a partir de clases ya existentes, estableciendo así una relación jerárquica entre ellas. Gracias a este proceso, una clase derivada o hija puede reutilizar los atributos y métodos definidos en una clase base o padre. Esto favorece la organización del código y evita repetir funcionalidades comunes.

Además, la clase base define una estructura general o comportamiento común que las clases derivadas comparten. A partir de esa base, cada clase hija puede especializar o adaptar dicho comportamiento, ya sea añadiendo nuevas características o redefiniendo métodos para ofrecer una implementación más específica según sus necesidades.

Este código no muestra herencia directamente, porque solo define una clase llamada Moto y no una clase hija que herede de ella. Sin embargo, sí puede entenderse como una clase base, ya que reúne atributos y métodos generales que podrían ser reutilizados por clases derivadas. La clase define información común de una moto, como modelo, potencia, consumo y kilometraje, además de métodos para mostrar sus especificaciones y calcular el consumo total. También utiliza una propiedad para controlar la actualización de los kilómetros, evitando asignar un valor menor al actual.

````python
class Moto1():
    """Esta clase representa un coche."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de instancia.

        Argumentos posicionales:
        modelo -- string que representa el modelo del coche
        potencia -- int que representa la potencia en cv
        consumo -- int que representa el consumo en litros/100km
        """
        self.modelo = modelo
        self.potencia = potencia
        self.consumo = consumo
        self._km_actuales = 0

    def especificaciones(self):
        """Muestra las especificaciones del coche."""
        print("Modelo:", self.modelo,
              "\nPotencia: {} cv".format(self.potencia),
              "\nConsumo: {} l/100km".format(self.consumo),
              "\nKilometros actuales:", self._km_actuales)

    @property
    def kilometros(self):
        return self._km_actuales

    @kilometros.setter
    def kilometros(self, kilometros):
        """Actualiza los kilometros actuales del coche."""
        if kilometros > self._km_actuales:
            self._km_actuales = kilometros
        else:
            print("ERROR: No se puede establecer un valor de km inferior al actual")

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} litros".format(consumo_total))
````

````python
mercedes = Moto1("mercedes c200", 180, 7)
````

````python
mercedes.especificaciones()
````

Desde la perspectiva de la herencia, esta clase podría servir como punto de partida para crear clases más específicas, como MotoDeportiva o MotoElectrica, que hereden esta estructura y añadan comportamientos propios.

**¿Qué sucede si quiero representar una moto electrica?**

````python
class Moto1():
    """Esta clase representa un coche."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de instancia.

        Argumentos posicionales:
        modelo -- string que representa el modelo del coche
        potencia -- int que representa la potencia en cv
        consumo -- int que representa el consumo
        """
        self.modelo = modelo
        self.potencia = potencia
        self.consumo = consumo
        self._km_actuales = 0
        self._combustible = "l/100km"

    def especificaciones(self):
        """Muestra las especificaciones del coche."""
        print("Modelo:", self.modelo,
              "\nPotencia: {} cv".format(self.potencia),
              "\nConsumo: {} {}".format(self.consumo, self._combustible),
              "\nKilometros actuales:", self._km_actuales)

    @property
    def kilometros(self):
        return self._km_actuales

    @kilometros.setter
    def kilometros(self, kilometros):
        """Actualiza los kilometros actuales del coche."""
        if kilometros > self._km_actuales:
            self._km_actuales = kilometros
        else:
            print("ERROR: No se puede establecer un valor de km inferior al actual")

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} litros".format(consumo_total))
````

````python
moto_electrica = Moto1("E1 model 3", 300, 15) # consumo: 15Kwh/100km
````

````python
moto_electrica.especificaciones()
````

En este punto podría definirse un método setter para modificar el valor del combustible; sin embargo, incorporar este tipo de lógica dentro de la clase base podría hacerla demasiado extensa, compleja y difícil de mantener. Cuando una clase concentra numerosos métodos y validaciones para atender casos particulares, aumenta el riesgo de generar código repetido y poco claro. Por esta razón, en situaciones como esta resulta más adecuado recurrir a la herencia, dejando en la clase base únicamente los elementos comunes y trasladando los comportamientos específicos a clases derivadas. De este modo, se consigue un diseño más ordenado, reutilizable y fácil de comprender.

````python
class Moto_electrica(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
````

````python
moto_E1 = Moto_electrica("E1 model 3", 300, 15)
````

````python
moto_E1.especificaciones()
````

### 10. Objetos dentro de una clase

En algunos casos, ciertas propiedades de una clase pueden tener la suficiente importancia o complejidad como para representarse mediante una clase propia. Cuando esto ocurre, en lugar de usar un valor simple, es posible asignar un objeto de esa segunda clase como atributo de la primera. Esto permite modelar la información de una forma más organizada, clara y cercana a la realidad.

````python
class Bateria:
    """Esta clase va a representar una bateria de un coche electrico."""

    def __init__(self, capacidad, tipo_pila, num_pilas, peso):
        self._capacidad = capacidad
        self._tipo_pila = tipo_pila
        self._num_pilas = num_pilas
        self._peso = peso

    def especificaciones(self):
        print("Capacidad {} kWh".format(self._capacidad),
             "\nTipo de pilas:", self._tipo_pila,
             "\nNumero de pilas:", self._num_pilas,
             "\nPeso de la bateria: {} kg".format(self._peso))
````

````python
bateria_tesla_modelS = Bateria(80, 2170, 203_136, 480)
````

````python
bateria_tesla_modelS.especificaciones()
````

````python
class CocheElectrico(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo, bateria):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
        self._bateria = bateria

    def detalles_bateria(self):
        """Muestra los detalles de la bateria del coche electrico."""
        self._bateria.especificaciones()

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} kWh".format(consumo_total))
````
